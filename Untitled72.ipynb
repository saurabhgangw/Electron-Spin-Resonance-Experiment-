```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Number of particles
num_particles = 100
# Radius of the spherical particles
radius = 1.0
# Desired density
desired_density = 0.64

# Calculate the volume required to achieve the desired density
volume_per_particle = (4/3) * np.pi * radius**3
total_volume_particles = num_particles * volume_per_particle
bounding_box_volume = total_volume_particles / desired_density
bounding_box_side_length = bounding_box_volume**(1/3)

# Generate initial random positions for particles within the calculated cube
initial_positions = np.random.rand(num_particles, 3) * bounding_box_side_length

# Define the objective function to minimize
def objective_function(positions):
    positions = positions.reshape((num_particles, 3))
    total_energy = 0.0

    for i in range(num_particles):
        for j in range(i + 1, num_particles):
            # Compute the distance between particles
            distance = np.linalg.norm(positions[i] - positions[j])
            # Compute the overlap (if any)
            overlap = 2 * radius - distance
            if overlap > 0:
                # Add a penalty for overlapping particles
                total_energy += overlap**2 #addding overlap (if any) to the variable 'total energy' every iteration of the for loop

    # # Add penalties for particles outside the bounding box
    # for i in range(num_particles):
    #     for d in range(3):  # Check x, y, z coordinates
    #         if positions[i, d] < 0:
    #             total_energy += (positions[i, d])**2
    #         elif positions[i, d] > bounding_box_side_length:
    #             total_energy += (positions[i, d] - bounding_box_side_length)**2

    return total_energy

# Define the gradient of the objective function
def gradient(positions):
    positions = positions.reshape((num_particles, 3))
    grad = np.zeros_like(positions)

    for i in range(num_particles):
        for j in range(num_particles):
            if i != j:
                # Compute the distance between particles
                distance = np.linalg.norm(positions[i] - positions[j])
                # Compute the overlap (if any)
                overlap = 2 * radius - distance
                if overlap > 0:
                    # Compute the gradient contribution for overlapping particles
                    grad_contribution = 2 * overlap * (positions[i] - positions[j]) / distance
                    grad[i] += grad_contribution

    # Add gradient contributions for particles outside the bounding box
    # for i in range(num_particles):
    #     for d in range(3):  # Check x, y, z coordinates
    #         if positions[i, d] < 0:
    #             grad[i, d] += 2 * positions[i, d]
    #         elif positions[i, d] > bounding_box_side_length:
    #             grad[i, d] += 2 * (positions[i, d] - bounding_box_side_length)

    return grad.flatten()

# Nonlinear Conjugate Gradient Method
def nonlinear_conjugate_gradient(objective_function, gradient, initial_positions, max_iter=1000, tol=1e-5):
    positions = initial_positions.flatten()
    r = -gradient(positions)
    p = r
    rsold = np.dot(r, r)

    for i in range(max_iter):
        Ap = gradient(positions + p * tol)
        alpha = rsold / np.dot(p, Ap)
        positions = positions + alpha * p
        r = r - alpha * Ap
        rsnew = np.dot(r, r)
        if np.sqrt(rsnew) < tol:
            break
        p = r + (rsnew / rsold) * p
        rsold = rsnew

    return positions

# Optimize particle positions using the Nonlinear Conjugate Gradient method
optimized_positions_flat = nonlinear_conjugate_gradient(objective_function, gradient, initial_positions)
optimized_positions = optimized_positions_flat.reshape((num_particles, 3))

# Plot initial and optimized positions
fig = plt.figure(figsize=(12, 6))

# Plot initial positions
ax1 = fig.add_subplot(121, projection='3d')
ax1.scatter(initial_positions[:, 0], initial_positions[:, 1], initial_positions[:, 2], color='blue')
ax1.set_title("Initial Positions")
ax1.set_xlim([0, bounding_box_side_length])
ax1.set_ylim([0, bounding_box_side_length])
ax1.set_zlim([0, bounding_box_side_length])

# Plot optimized positions
ax2 = fig.add_subplot(122, projection='3d')
ax2.scatter(optimized_positions[:, 0], optimized_positions[:, 1], optimized_positions[:, 2], color='red')
ax2.set_title("Optimized Positions")
ax2.set_xlim([0, bounding_box_side_length])
ax2.set_ylim([0, bounding_box_side_length])
ax2.set_zlim([0, bounding_box_side_length])

plt.show()

# Calculate final density
final_volume_particles = num_particles * volume_per_particle
final_bounding_box_volume = bounding_box_side_length**3
final_density = final_volume_particles / final_bounding_box_volume

print(f"Final density: {final_density:.2f}")
```
